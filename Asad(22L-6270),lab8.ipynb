```python
# Import required libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load the dataset
df = pd.read_csv("cancer.csv")

# Display first 5 rows
print(df.head())

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Define features (X) and target (y)
x = df.drop(columns=['Diagnosis'], axis=1)   # Remove target column
y = df['Diagnosis']                           # Target column

# Split data: 75% train, 25% test
x_trainn, x_testt, y_trainn, y_testt = train_test_split(
    x, y, test_size=0.25, random_state=42
)

# ========================================
# 1. NAIVE BAYES
# ========================================
print("\n" + "="*50)
print("NAIVE BAYES")
print("="*50)

model_nb = GaussianNB()
model_nb.fit(x_trainn, y_trainn)
y_pred_nb = model_nb.predict(x_testt)

nb_acc = accuracy_score(y_testt, y_pred_nb)
print(f"Accuracy: {nb_acc:.4f}")

# ========================================
# 2. KNN (k=3)
# ========================================
print("\n" + "="*50)
print("KNN (k=3)")
print("="*50)

model_kk = KNeighborsClassifier(n_neighbors=3)
model_kk.fit(x_trainn, y_trainn)
y_pred_kk = model_kk.predict(x_testt)

kk_acc = accuracy_score(y_testt, y_pred_kk)
print(f"Accuracy: {kk_acc:.4f}")

# ========================================
# 3. DECISION TREE
# ========================================
print("\n" + "="*50)
print("DECISION TREE")
print("="*50)

model_dt = DecisionTreeClassifier(random_state=0)
model_dt.fit(x_trainn, y_trainn)
y_pred_dt = model_dt.predict(x_testt)

dt_acc = accuracy_score(y_testt, y_pred_dt)
print(f"Accuracy: {dt_acc:.4f}")

# ========================================
# FINAL COMPARISON
# ========================================
print("\n" + "="*50)
print("ACCURACY SUMMARY")
print("="*50)
print(f"Naive Bayes     : {nb_acc:.4f}")
print(f"KNN (k=3)       : {kk_acc:.4f}")
print(f"Decision Tree   : {dt_acc:.4f}")

# Find best model
accuracies = {
    "Naive Bayes": nb_acc,
    "KNN": kk_acc,
    "Decision Tree": dt_acc
}
best_model = max(accuracies, key=accuracies.get)
print(f"\nBest Model: {best_model} with Accuracy: {accuracies[best_model]:.4f}")
```

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB 
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Training Data
train_data = {
    "x1": [60, 62, 67, 70, 71, 72, 75, 78],
    "x2": [22, 25, 24, 20, 15, 14, 14, 11],
    "y":  [0,  0,  0,  0,  1,  1,  1,  1]
}

# Test Data
test_data = {
    "x1": [61, 71, 73, 79],
    "x2": [23, 19, 15, 13],
    "y":  [0,  0,  1,  1]
}

# Convert to DataFrame
df_train = pd.DataFrame(train_data)
df_test  = pd.DataFrame(test_data)

# Features and Labels
x_train = df_train[["x1", "x2"]]
y_train = df_train["y"]           # Single column → use .ravel()
x_test  = df_test[["x1", "x2"]]
y_test  = df_test["y"]

# ========================================
# Define Models Function
# ========================================
def models(x_train, y_train, x_test, y_test):
    
    # Fix: Flatten y_train/y_test if needed (avoids warning)
    y_train = y_train.values.ravel() if hasattr(y_train, 'values') else y_train
    y_test  = y_test.values.ravel()  if hasattr(y_test, 'values') else y_test

    # -------------------------------
    # 1. RANDOM FOREST
    # -------------------------------
    print("RANDOM FOREST")
    rf = RandomForestClassifier(n_estimators=10, random_state=0)
    rf.fit(x_train, y_train)
    y_pred_rf = rf.predict(x_test)                    # Fixed: rf.predict, not re.predict
    acc_rf = accuracy_score(y_test, y_pred_rf)        # Fixed: use y_pred_rf
    print("Accuracy:", acc_rf)
    print()

    # -------------------------------
    # 2. NAIVE BAYES
    # -------------------------------
    print("NAIVE BAYES")
    model_nb = GaussianNB()
    model_nb.fit(x_train, y_train)
    y_pred_nb = model_nb.predict(x_test)
    nb_acc = accuracy_score(y_test, y_pred_nb)
    print("Accuracy:", nb_acc)
    print()

    # -------------------------------
    # 3. KNN
    # -------------------------------
    print("KNN (k=3)")
    model_kk = KNeighborsClassifier(n_neighbors=3)
    model_kk.fit(x_train, y_train)
    y_pred_kk = model_kk.predict(x_test)
    kk_acc = accuracy_score(y_test, y_pred_kk)
    print("Accuracy:", kk_acc)
    print()

    # -------------------------------
    # 4. DECISION TREE
    # -------------------------------
    print("DECISION TREE")
    model_dt = DecisionTreeClassifier(random_state=0)
    model_dt.fit(x_train, y_train)
    y_pred_dt = model_dt.predict(x_test)
    dt_acc = accuracy_score(y_test, y_pred_dt)
    print("Accuracy:", dt_acc)
    print()

    # -------------------------------
    # SUMMARY
    # -------------------------------
    print("="*50)
    print("FINAL ACCURACY SUMMARY")
    print("="*50)
    print(f"Random Forest   : {acc_rf:.4f}")
    print(f"Naive Bayes     : {nb_acc:.4f}")
    print(f"KNN (k=3)       : {kk_acc:.4f}")
    print(f"Decision Tree   : {dt_acc:.4f}")

    # Best Model
    scores = {"RF": acc_rf, "NB": nb_acc, "KNN": kk_acc, "DT": dt_acc}
    best = max(scores, key=scores.get)
    print(f"\nBest Model: {best} → {scores[best]:.4f}")

# ========================================
# Run Models
# ========================================
models(x_train, y_train, x_test, y_test)
```
