```python
import pandas as pd
import numpy as np

# Data from previous input
data1 = pd.DataFrame({
    'X': [0, 0.5, 1, 1.5, 2, 2.5, 3, 3.5],
    'Y': [2, 1.5, 1.5, 1, 1, 0, -0.5, -2]
})
X = data1['X'].values
Y = data1['Y'].values
m = len(Y)

def predict(X, theta0, theta1):
    return theta0 + theta1 * X

def gradient_descent_mse(X, Y, lr=0.02, iterations=10):
    theta0, theta1 = 0.0, 1.0  # Initial parameters
    for i in range(iterations):
        y_pred = predict(X, theta0, theta1)
        error = y_pred - Y

        # Gradients for MSE
        grad0 = (1/m) * np.sum(error)
        grad1 = (1/m) * np.sum(error * X)

        # Update parameters
        theta0 = theta0 - lr * grad0
        theta1 = theta1 - lr * grad1

        # Compute MSE for monitoring
       # mse = (1/m) * np.sum(error**2)
       # print(f"MSE Iteration {i+1}: theta0 = {theta0:.4f}, theta1 = {theta1:.4f}, MSE = {mse:.4f}")

    return theta0, theta1

def gradient_descent_mae(X, Y, lr=0.02, iterations=10):
    theta0, theta1 = 0.0, 1.0  # Initial parameters
    for i in range(iterations):
        y_pred = predict(X, theta0, theta1)
        error = y_pred - Y

        # Gradients for MAE
        sign = np.sign(error)
        grad0 = (1/m) * np.sum(sign)
        grad1 = (1/m) * np.sum(sign * X)

        # Update parameters
        theta0 = theta0 - lr * grad0
        theta1 = theta1 - lr * grad1

        # Compute MAE for monitoring
       # mae = (1/m) * np.sum(np.abs(error))
      #  print(f"MAE Iteration {i+1}: theta0 = {theta0:.4f}, theta1 = {theta1:.4f}, MAE = {mae:.4f}")

    return theta0, theta1

# Run both optimizations
theta0_mae, theta1_mae = gradient_descent_mae(X, Y, lr=0.02, iterations=1000)
theta0_mse, theta1_mse = gradient_descent_mse(X, Y, lr=0.02, iterations=1000)

# Print final parameters
print("\nFinal Parameters:")
print(f"Theta0 (MAE): {theta0_mae:.4f}")
print(f"Theta1 (MAE): {theta1_mae:.4f}")
print(f"Theta0 (MSE): {theta0_mse:.4f}")
print(f"Theta1 (MSE): {theta1_mse:.4f}")
```

```python
import numpy as np
import pandas as pd

# Data from the provided input
data1 = pd.DataFrame({
    'X1': [1, 3.5, 0.75, 2.5, 1.5],
    'X2': [2, 1.5, 1.25, 2, 0.5],
    'Y': [2, 2.5, 1.5, 3.25, 1]
})

X = data1[['X1', 'X2']].values  # Using both X1 and X2 as features
Y = data1['Y'].values
m = len(Y)  # Number of samples

def predict(X, theta0, theta1, theta2):
    return theta0 + theta1 * X[:, 0] + theta2 * X[:, 1]

def gradient_descent_mse(X, Y, lr=0.01, iterations=1000):
    theta0, theta1, theta2 = 0.0, 0.0, 0.0  # Initial parameters
    for i in range(iterations):
        y_pred = predict(X, theta0, theta1, theta2)
        error = y_pred - Y

        # Gradients for MSE
        grad0 = (1/m) * np.sum(error)
        grad1 = (1/m) * np.sum(error * X[:, 0])
        grad2 = (1/m) * np.sum(error * X[:, 1])

        # Update parameters
        theta0 = theta0 - lr * grad0
        theta1 = theta1 - lr * grad1
        theta2 = theta2 - lr * grad2

    return theta0, theta1, theta2

def gradient_descent_mae(X, Y, lr=0.01, iterations=1000):
    theta0, theta1, theta2 = 0.0, 0.0, 0.0  # Initial parameters
    for i in range(iterations):
        y_pred = predict(X, theta0, theta1, theta2)
        error = y_pred - Y

        # Gradients for MAE
        sign = np.sign(error)
        grad0 = (1/m) * np.sum(sign)
        grad1 = (1/m) * np.sum(sign * X[:, 0])
        grad2 = (1/m) * np.sum(sign * X[:, 1])

        # Update parameters
        theta0 = theta0 - lr * grad0
        theta1 = theta1 - lr * grad1
        theta2 = theta2 - lr * grad2

    return theta0, theta1, theta2

# Run both optimizations
theta0_mae, theta1_mae, theta2_mae = gradient_descent_mae(X, Y, lr=0.01, iterations=10000)
theta0_mse, theta1_mse, theta2_mse = gradient_descent_mse(X, Y, lr=0.01, iterations=10000)

# Print final parameters
print("\nFinal Parameters:")
print(f"Theta0 (MAE): {theta0_mae:.4f}")
print(f"Theta1 (MAE): {theta1_mae:.4f} (for X1)")
print(f"Theta2 (MAE): {theta2_mae:.4f} (for X2)")
print(f"Theta0 (MSE): {theta0_mse:.4f}")
print(f"Theta1 (MSE): {theta1_mse:.4f} (for X1)")
print(f"Theta2 (MSE): {theta2_mse:.4f} (for X2)")
```
