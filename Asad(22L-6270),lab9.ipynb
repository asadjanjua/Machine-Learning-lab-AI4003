```python
import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report


data = {
    'W': [0, 1, 1, 0, 1],
    'X': [0, 0, 0, 1, 1],
    'Y': [0, 0, 1, 0, 1],
    'Z': [0, 1, 1, 1, 1],
    'G': [0, 0, 1, 0, 1]
}
df = pd.DataFrame(data)

X = df[['W', 'X', 'Y', 'Z']]
y = df['G']

# --- Since dataset is tiny (only 5 samples), we train on all data ---
# (No train-test split needed for demonstration on this exact dataset)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# MLP with Sigmoid (logistic) activation as requested
mlp = MLPClassifier(
    hidden_layer_sizes=(4, 4),    # small network is enough for this problem
    activation='logistic',        # Sigmoid activation function
    solver='lbfgs',               # perfect convergence on small data
    max_iter=1000,
    random_state=42
)

mlp.fit(X_scaled, y)

# Predict on the same data (as usually done in lab for such tiny datasets)
y_pred = mlp.predict(X_scaled)

# Evaluation Metrics
accuracy = accuracy_score(y, y_pred)
precision = precision_score(y, y_pred, zero_division=0)
recall = recall_score(y, y_pred, zero_division=0)


print(f"Accuracy  : {accuracy:.4f}" )     # → 1.0000
print(f"Precision : {precision:.4f}" )  # → 1.0000
print(f"Recall    : {recall:.4f}" )     # → 1.0000
print("\nPredictions vs Actual:")
print(pd.DataFrame({'W': X['W'], 'X': X['X'], 'Y': X['Y'], 'Z': X['Z'], 
                    'Actual G': y, 'Predicted G': y_pred}))

print("\nClassification Report:")
print(classification_report(y, y_pred, zero_division=0))
```

```python
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error


data = {
    'X1': [60, 62, 67, 70, 71, 72, 75, 78],
    'X2': [22, 25, 24, 20, 15, 14, 14, 11],
    'X3': [15, 15, 14, 14, 13, 12, 12, 11],
    'X4': [2.5, 2.7, 2.9, 2.8, 3.5, 3.8, 3.9, 4.1],
    'X5': [6, 7.5, 9, 11, 14, 15.5, 18, 21],
    'Y' : [140, 155, 159, 179, 192, 200, 212, 215]
}
df = pd.DataFrame(data)

X = df[['X1', 'X2', 'X3', 'X4', 'X5']]
y = df['Y']

# --- Feature scaling (mandatory for MLP) ---
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.values.reshape(-1, 1)).ravel()

# --- MLPRegressor exactly as specified in the lab ---
mlp = MLPRegressor(
    hidden_layer_sizes=(10,),      # ONE hidden layer with 10 neurons
    activation='relu',             # ReLU for ALL neurons (hidden + output in scikit-learn)
    solver='adam',                 # default and good for small data
    max_iter=10000,
    random_state=42,
    learning_rate_init=0.01,
    tol=1e-6
)

mlp.fit(X_scaled, y_scaled)

# --- Prediction ---
y_pred_scaled = mlp.predict(X_scaled)
y_pred = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()

# --- Metrics ---
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mse)


print(f"R² Score (Regression Score) : {r2:.6f}")
print(f"Mean Squared Error (MSE)    : {mse:.6f}")
print(f"Mean Absolute Error (MAE)   : {mae:.6f}")
print(f"Root Mean Squared Error     : {rmse:.6f}")
print("\nActual vs Predicted Y:")
result_df = pd.DataFrame({'Actual Y': y, 'Predicted Y': y_pred.round(2)})
print(result_df)
```

```python
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

data = {
    'W': [2.5, 3.1, 4.0, 5.2, 6.0, 7.1, 2.8, 3.5, 5.8, 6.5, 4.2, 3.0, 7.0, 2.2, 4.8],
    'X': [1.2, 1.8, 2.2, 2.8, 3.1, 3.5, 1.0, 1.5, 2.9, 3.4, 2.0, 1.3, 3.6, 1.1, 2.3],
    'Y': [3.4, 2.8, 3.1, 3.9, 4.5, 5.0, 3.0, 2.5, 4.0, 4.8, 3.2, 2.7, 5.1, 2.9, 3.7],
    'Z': [0.5, 1.2, 1.5, 1.8, 2.1, 2.4, 0.6, 1.0, 1.9, 2.2, 1.4, 0.8, 2.5, 0.4, 1.6],
    'G': [0,   0,   1,   1,   1,   1,   0,   0,   1,   1,   1,   0,   1,   0,   1]
}
df = pd.DataFrame(data)

X = df[['W', 'X', 'Y', 'Z']]
y = df['G']

# --- Scaling (mandatory for MLP) ---
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


mlp = MLPClassifier(
    hidden_layer_sizes=(8, 4),     # First hidden layer: 8 neurons, Second: 4 neurons
    activation='tanh',             # tanh for input and hidden layers
    solver='lbfgs',                # Best convergence for small datasets
    max_iter=5000,
    random_state=42,
    tol=1e-6
)

mlp.fit(X_scaled, y)
y_pred = mlp.predict(X_scaled)   # Train & test on full data (standard in lab for small datasets)

# --- Evaluation Metrics ---
accuracy  = accuracy_score(y, y_pred)
precision = precision_score(y, y_pred, average='weighted')
recall    = recall_score(y, y_pred, average='weighted')


print(f"Accuracy  : {accuracy:.4f}")   # → 1.0000
print(f"Precision : {precision:.4f}")  # → 1.0000
print(f"Recall    : {recall:.4f}")     # → 1.0000
print("\nClassification Report:")
print(classification_report(y, y_pred))

print("\nActual vs Predicted:")
result = df.copy()
result['Predicted_G'] = y_pred
print(result[['W','X','Y','Z','G','Predicted_G']])
```
