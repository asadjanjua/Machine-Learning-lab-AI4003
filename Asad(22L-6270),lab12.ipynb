```python
# ==============================================================
# EXPERIMENT 12: CNN for Image Classification (CIFAR-10 + Fashion-MNIST)
# ==============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, datasets
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# --------------------------------------------------------------
# Part 1: CNN on CIFAR-10 (as given in the lab procedure)
# --------------------------------------------------------------

print("=== CIFAR-10 CNN with StandardScaler Preprocessing ===\n")

# Step 1: Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Step 2: StandardScaler preprocessing (reshape → scale → reshape back)
# Reshape to 2D for StandardScaler
x_train_reshaped = x_train.reshape((x_train.shape[0], -1))  # (50000, 32*32*3)
x_test_reshaped = x_test.reshape((x_test.shape[0], -1))     # (10000, 32*32*3)

# Fit scaler on training data only and transform both sets
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train_reshaped)
x_test_scaled = scaler.transform(x_test_reshaped)

# Reshape back to (samples, 32, 32, 3)
x_train = x_train_scaled.reshape(x_train.shape)
x_test = x_test_scaled.reshape(x_test.shape)

# One-hot encode labels
num_classes = 10
y_train = tf.keras.utils.to_categorical(y_train, num_classes)
y_test = tf.keras.utils.to_categorical(y_test, num_classes)

print(f"Training data shape (after scaling): {x_train.shape}")
print(f"Test data shape (after scaling):     {x_test.shape}\n")

# Step 3: Define CNN model
model = models.Sequential([
    # First Conv Block
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),

    # Second Conv Block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Classifier
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.summary()

# Step 4: Compile and train
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train,
                    epochs=10,
                    batch_size=64,
                    validation_split=0.1,
                    verbose=1)

# Step 5: Evaluate on test set
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"\nCIFAR-10 Test Accuracy: {test_acc*100:.2f}%")
print(f"CIFAR-10 Test Loss: {test_loss:.4f}\n")

# --------------------------------------------------------------
# In-Lab Exercise 3: CNN on Fashion-MNIST with StandardScaler
# --------------------------------------------------------------

print("="*60)
print("=== Exercise 3: Fashion-MNIST CNN with StandardScaler ===")
print("="*60)

# Load Fashion-MNIST
(x_train, y_train), (x_test, y_test) = datasets.fashion_mnist.load_data()

# Add channel dimension: (28, 28) → (28, 28, 1)
x_train = x_train.reshape((x_train.shape[0], 28, 28, 1))
x_test = x_test.reshape((x_test.shape[0], 28, 28, 1))

# StandardScaler preprocessing (reshape → scale → reshape back)
x_train_reshaped = x_train.reshape((x_train.shape[0], -1))  # (60000, 784)
x_test_reshaped = x_test.reshape((x_test.shape[0], -1))     # (10000, 784)

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train_reshaped)
x_test_scaled = scaler.transform(x_test_reshaped)

# Reshape back to image format with channel
x_train = x_train_scaled.reshape((-1, 28, 28, 1))
x_test = x_test_scaled.reshape((-1, 28, 28, 1))

# One-hot encode labels
y_train_cat = tf.keras.utils.to_categorical(y_train, 10)
y_test_cat = tf.keras.utils.to_categorical(y_test, 10)

print(f"Fashion-MNIST Training data shape (after scaling): {x_train.shape}")
print(f"Fashion-MNIST Test data shape (after scaling):     {x_test.shape}\n")

# Define two-block CNN for Fashion-MNIST
fashion_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1), padding='same'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

fashion_model.summary()

# Compile
fashion_model.compile(optimizer='adam',
                      loss='categorical_crossentropy',
                      metrics=['accuracy'])

# Train
history_fashion = fashion_model.fit(x_train, y_train_cat,
                                    epochs=10,
                                    batch_size=64,
                                    validation_split=0.1,
                                    verbose=1)

# Evaluate
test_loss_f, test_acc_f = fashion_model.evaluate(x_test, y_test_cat, verbose=0)
print(f"\nFashion-MNIST Test Accuracy: {test_acc_f*100:.2f}%")
print(f"Fashion-MNIST Test Loss: {test_loss_f:.4f}")
```

```python
# ==============================================================
# Exercise 1: Perceptron with K-Fold CV on Breast Cancer Dataset
# ==============================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# -------------------------- Load Breast Cancer Dataset --------------------------
# Option 1: If you have a CSV file named 'cancer.csv', 'breast_cancer.csv', etc.
df = pd.read_csv('cancer (1) (2) (1).csv')  # ← Change filename if needed

# Option 2: Use the built-in Scikit-learn Breast Cancer dataset (Recommended & Most Common in Labs)
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target  # 0 = malignant, 1 = benign

print("Dataset loaded: Breast Cancer Wisconsin")
print(f"Shape: {df.shape}")
print(f"Classes: {np.unique(df['target'])} → 0: Malignant, 1: Benign\n")

# -------------------------- Prepare Features & Labels --------------------------
X = df.drop('target', axis=1)
y = df['target']

# Scale features (Very Important for Perceptron!)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# -------------------------- Train-Test Split 70:30 --------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)

# -------------------------- Perceptron Model --------------------------
perceptron = Perceptron(max_iter=1000, eta0=1.0, random_state=42, tol=1e-3)

# -------------------------- Metrics Function --------------------------
def compute_metrics(y_true, y_pred):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, average='macro', zero_division=0),
        'Recall': recall_score(y_true, y_pred, average='macro', zero_division=0),
        'F1-Score': f1_score(y_true, y_pred, average='macro', zero_division=0)
    }

# Train once on full training set
perceptron.fit(X_train, y_train)

# Training metrics
y_train_pred = perceptron.predict(X_train)
train_metrics = compute_metrics(y_train, y_train_pred)

# Test metrics
y_test_pred = perceptron.predict(X_test)
test_metrics = compute_metrics(y_test, y_test_pred)

# -------------------------- K-Fold Cross Validation (K=5 & K=10) --------------------------
results = []

for K in [5, 10]:
    kf = KFold(n_splits=K, shuffle=True, random_state=42)

    cv_acc = cross_val_score(perceptron, X_train, y_train, cv=kf, scoring='accuracy')
    cv_prec = cross_val_score(perceptron, X_train, y_train, cv=kf, scoring='precision_macro')
    cv_rec = cross_val_score(perceptron, X_train, y_train, cv=kf, scoring='recall_macro')
    cv_f1 = cross_val_score(perceptron, X_train, y_train, cv=kf, scoring='f1_macro')

    results.append({
        'K': K,
        'Train_Acc': round(train_metrics['Accuracy'], 4),
        'Val_Acc': round(cv_acc.mean(), 4),
        'Test_Acc': round(test_metrics['Accuracy'], 4),
        'Train_Prec': round(train_metrics['Precision'], 4),
        'Val_Prec': round(cv_prec.mean(), 4),
        'Test_Prec': round(test_metrics['Precision'], 4),
        'Train_Rec': round(train_metrics['Recall'], 4),
        'Val_Rec': round(cv_rec.mean(), 4),
        'Test_Rec': round(test_metrics['Recall'], 4),
        'Train_F1': round(train_metrics['F1-Score'], 4),
        'Val_F1': round(cv_f1.mean(), 4),
        'Test_F1': round(test_metrics['F1-Score'], 4),
    })

# -------------------------- Final Table Output (Exact Lab Format) --------------------------
print("\n" + "="*90)
print("                   PERCEPTRON MODEL PERFORMANCE (Breast Cancer Dataset)")
print("="*90)
print(f"{'Metric':<10} {'K':<3} {'Training Data':<16} {'Validation Data':<18} {'Test Data':<12}")
print("-"*90)

metrics_order = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

for metric in metrics_order:
    for res in results:
        if metric == 'Accuracy':
            print(f"{metric:<10} {res['K']:<3} {res['Train_Acc']:<16} {res['Val_Acc']:<18} {res['Test_Acc']:<12}")
        elif metric == 'Precision':
            print(f"{metric:<10} {res['K']:<3} {res['Train_Prec']:<16} {res['Val_Prec']:<18} {res['Test_Prec']:<12}")
        elif metric == 'Recall':
            print(f"{metric:<10} {res['K']:<3} {res['Train_Rec']:<16} {res['Val_Rec']:<18} {res['Test_Rec']:<12}")
        elif metric == 'F1-Score':
            print(f"{metric:<10} {res['K']:<3} {res['Train_F1']:<16} {res['Val_F1']:<18} {res['Test_F1']:<12}")
    print()  # Blank line between metrics
```

```python
# ==============================================================
# Exercise 2: Logistic Regression + Comparison with Perceptron
# Breast Cancer Dataset
# ==============================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import Perceptron, LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.datasets import load_breast_cancer

# -------------------------- Load Dataset --------------------------
data = load_breast_cancer()
X = data.data
y = data.target

# Feature scaling (critical for both models)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split 70:30
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)

# -------------------------- Models --------------------------
perceptron = Perceptron(max_iter=1000, eta0=1.0, random_state=42, tol=1e-3)
logreg = LogisticRegression(max_iter=1000, random_state=42)

# -------------------------- Metrics Function --------------------------
def get_metrics(y_true, y_pred):
    return {
        'Accuracy': round(accuracy_score(y_true, y_pred), 4),
        'Precision': round(precision_score(y_true, y_pred, average='macro'), 4),
        'Recall': round(recall_score(y_true, y_pred, average='macro'), 4),
        'F1-Score': round(f1_score(y_true, y_pred, average='macro'), 4)
    }

# -------------------------- Train & Evaluate Both Models --------------------------
models = {'Perceptron': perceptron, 'Logistic Regression': logreg}
results_all = []

for name, model in models.items():
    # Train
    model.fit(X_train, y_train)

    # Training metrics
    train_pred = model.predict(X_train)
    train_metrics = get_metrics(y_train, train_pred)

    # Test metrics
    test_pred = model.predict(X_test)
    test_metrics = get_metrics(y_test, test_pred)

    # K-Fold CV on training data
    cv_results = []
    for K in [5, 10]:
        kf = KFold(n_splits=K, shuffle=True, random_state=42)

        acc = cross_val_score(model, X_train, y_train, cv=kf, scoring='accuracy').mean()
        prec = cross_val_score(model, X_train, y_train, cv=kf, scoring='precision_macro').mean()
        rec = cross_val_score(model, X_train, y_train, cv=kf, scoring='recall_macro').mean()
        f1 = cross_val_score(model, X_train, y_train, cv=kf, scoring='f1_macro').mean()

        cv_results.append({
            'K': K,
            'Val_Acc': round(acc, 4),
            'Val_Prec': round(prec, 4),
            'Val_Rec': round(rec, 4),
            'Val_F1': round(f1, 4)
        })

    # Store results for this model
    for cv in cv_results:
        results_all.append({
            'Model': name,
            'K': cv['K'],
            'Train_Acc': train_metrics['Accuracy'],
            'Val_Acc': cv['Val_Acc'],
            'Test_Acc': test_metrics['Accuracy'],
            'Train_Prec': train_metrics['Precision'],
            'Val_Prec': cv['Val_Prec'],
            'Test_Prec': test_metrics['Precision'],
            'Train_Rec': train_metrics['Recall'],
            'Val_Rec': cv['Val_Rec'],
            'Test_Rec': test_metrics['Recall'],
            'Train_F1': train_metrics['F1-Score'],
            'Val_F1': cv['Val_F1'],
            'Test_F1': test_metrics['F1-Score'],
        })

# -------------------------- Final Comparison Table --------------------------
print("\n" + "="*110)
print("          COMPARISON: PERCEPTRON vs LOGISTIC REGRESSION (Breast Cancer Dataset)")
print("="*110)
print(f"{'Metric':<10} {'K':<3} {'Model':<20} {'Training':<12} {'Validation':<14} {'Test':<10}")
print("-"*110)

metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

for metric in metrics:
    for res in results_all:
        if metric == 'Accuracy':
            print(f"{metric:<10} {res['K']:<3} {res['Model']:<20} {res['Train_Acc']:<12} {res['Val_Acc']:<14} {res['Test_Acc']:<10}")
        elif metric == 'Precision':
            print(f"{metric:<10} {res['K']:<3} {res['Model']:<20} {res['Train_Prec']:<12} {res['Val_Prec']:<14} {res['Test_Prec']:<10}")
        elif metric == 'Recall':
            print(f"{metric:<10} {res['K']:<3} {res['Model']:<20} {res['Train_Rec']:<12} {res['Val_Rec']:<14} {res['Test_Rec']:<10}")
        elif metric == 'F1-Score':
            print(f"{metric:<10} {res['K']:<3} {res['Model']:<20} {res['Train_F1']:<12} {res['Val_F1']:<14} {res['Test_F1']:<10}")
    print()  # Blank line between metrics
```
