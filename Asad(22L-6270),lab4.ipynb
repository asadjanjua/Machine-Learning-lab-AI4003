```python
import numpy as np
import pandas as pd

# Data
data = {
    'X': [0, 0.5, 1, 1.5, 2, 2.5, 3, 3.5],
    'Y': [2, 1.5, 1.5, 1, 1, 0, -0.5, -2]
}

# Create DataFrame
df = pd.DataFrame(data)

# Closed-form solution for linear regression
X = df['X'].values
Y = df['Y'].values
n = len(X)

# Calculate slope (m) and intercept (b)
X_mean = np.mean(X)
Y_mean = np.mean(Y)
m = np.sum((X - X_mean) * (Y - Y_mean)) / np.sum((X - X_mean) ** 2)
b = Y_mean - m * X_mean

# Predicted values
Y_pred = m * X + b

# Coefficient of determination (R²)
SS_tot = np.sum((Y - Y_mean) ** 2)
SS_res = np.sum((Y - Y_pred) ** 2)
R_squared = 1 - (SS_res / SS_tot)

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")
print(f"Coefficient of Determination (R²): {R_squared:.4f}")
```

```python
import numpy as np
import pandas as pd

# Data
data = {
    'X1': [60, 62, 67, 70, 71, 72, 75, 78],
    'X2': [22, 25, 24, 20, 15, 14, 14, 11],
    'Y': [140, 155, 159, 179, 192, 200, 212, 215]
}

# Create DataFrame
df = pd.DataFrame(data)

# Convert to numpy arrays
X1 = df['X1'].values
X2 = df['X2'].values
Y = df['Y'].values
n = len(Y)

# Add bias term and create feature matrix
X = np.column_stack((np.ones(n), X1, X2))
Y = Y.reshape(-1, 1)

# Closed-form solution for multiple linear regression (Normal Equation)
theta = np.dot(np.dot(np.linalg.inv(np.dot(X.T, X)), X.T), Y)

# Extract parameters
theta0, theta1, theta2 = theta.flatten()

# Predicted values
Y_pred = theta0 + theta1 * X1 + theta2 * X2

# Coefficient of determination (R²)
Y_mean = np.mean(Y)
SS_tot = np.sum((Y - Y_mean) ** 2)
SS_res = np.sum((Y - Y_pred.reshape(-1, 1)) ** 2)
R_squared = 1 - (SS_res / SS_tot)

# Print slope, intercept, and R²
print(f"Intercept: {theta0:.4f}")
print(f"Slope for X1: {theta1:.4f}")
print(f"Slope for X2: {theta2:.4f}")
print(f"R²: {R_squared:.4f}")
```

```python


import numpy as np
import pandas as pd



data1=pd.DataFrame({
"X1":[60,62,67,70,71,72,75,78],
"X2":[22,25,24,20,15,14,14,11],
"Y":[140,155,159,179,192,200,212,215]
})
X1=data1["X1"].values
X2=data1["X2"].values
Y=data1["Y"].values
m=len(Y)


# Prediction function
def predict(X1, X2, theta0, theta1, theta2):
    return theta0 + theta1 * X1 + theta2 * X2

# Gradient descent with MSE
def gradient_descent_mse(X1, X2, Y, lr=0.0001, iterations=1000):
    theta0, theta1, theta2 = 0.0, 0.0, 0.0
    for _ in range(iterations):
        pred = predict(X1, X2, theta0, theta1, theta2)
        error = pred - Y
        grad0 = (1/m) * np.sum(error)
        grad1 = (1/m) * np.sum(error * X1)
        grad2 = (1/m) * np.sum(error * X2)
        theta0 -= lr * grad0
        theta1 -= lr * grad1
        theta2 -= lr * grad2
    return theta0, theta1, theta2

# Run gradient descent
theta0_mse, theta1_mse, theta2_mse = gradient_descent_mse(X1, X2, Y, lr=0.0001, iterations=1000)

# Print parameters
print("MSE Results after 1000 iterations:")
print(f"theta0 = {round(theta0_mse, 4)}, theta1 = {round(theta1_mse, 4)}, theta2 = {round(theta2_mse, 4)}")

# Compute R^2
predictions = predict(X1, X2, theta0_mse, theta1_mse, theta2_mse)
y_mean = np.mean(Y)
SS_res = np.sum((Y - predictions) ** 2)
SS_tot = np.sum((Y - y_mean) ** 2)
r2 = 1 - (SS_res / SS_tot)
print("R^2 Score:", round(r2, 4))
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Data
data1 = pd.DataFrame({
    "X1": [60, 62, 67, 70, 71, 72, 75, 78],
    "X2": [22, 25, 24, 20, 15, 14, 14, 11],
    "Y": [140, 155, 159, 179, 192, 200, 212, 215]
})

# Prepare features and target
X = data1[["X1", "X2"]].values
Y = data1["Y"].values

# Create and fit the model
model = LinearRegression()
model.fit(X, Y)

# Extract parameters
intercept_sklearn = model.intercept_
slope1_sklearn = model.coef_[0]
slope2_sklearn = model.coef_[1]

# Predictions
Y_pred_sklearn = model.predict(X)

# R² Score
r2_sklearn = r2_score(Y, Y_pred_sklearn)

# Print results
print("Scikit-Learn Results:")
print(f"Intercept = {intercept_sklearn:.4f}")
print(f"Slope for X1 = {slope1_sklearn:.4f}")
print(f"Slope for X2 = {slope2_sklearn:.4f}")
print(f"R² Score = {r2_sklearn:.4f}")
```
