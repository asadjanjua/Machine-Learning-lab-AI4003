```python
import pandas as pd
import numpy as np

# Training and Test data as DataFrame
train_data = {
    "x1": [60, 62, 67, 70, 71, 72, 75, 78],
    "x2": [22, 25, 24, 20, 15, 14, 14, 11],
    "y": [0, 0, 0, 0, 1, 1, 1, 1]
}
test_data = {
    "X1": [61, 71, 73, 79],
    "X2": [23, 19, 15, 13],
    "Y": [0, 0, 1, 1]
}

df_train = pd.DataFrame(train_data)
df_test = pd.DataFrame(test_data)
X_train = np.c_[np.ones(len(df_train)), df_train[["x1", "x2"]].values]  # Add bias term
y_train = df_train["y"].values
X_test = np.c_[np.ones(len(df_test)), df_test[["X1", "X2"]].values]    # Add bias term
y_test = df_test["Y"].values

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def cost_function(y, y_pred):
    m = len(y)
    epsilon = 1e-15
    cost = -(1 / m) * np.sum(y * np.log(y_pred + epsilon) + (1 - y) * np.log(1 - y_pred + epsilon))
    return cost

def gradient_descent(X, y, lr=0.02, iterations=10000):
    theta = np.zeros(X.shape[1])  # Initialize theta for bias and features
    n = len(y)
    cost_history = []
    for it in range(iterations):
        z = np.dot(X, theta)
        y_pred = sigmoid(z)
        cost = cost_function(y, y_pred)
        cost_history.append(cost)
        error = y_pred - y
        gradient = (1 / n) * np.dot(X.T, error)
        theta = theta - lr * gradient
        if it % 2000 == 0:
            print(f"iteration {it}, Cost: {cost:.6f}")
    return theta, cost_history

def predict(X, theta):
    predict = sigmoid(np.dot(X, theta))
    predictions = []
    for p in predict:
        if p >= 0.5:
            predictions.append(1)
        else:
            predictions.append(0)
    return np.array(predictions), predict

def evaluate_model(y, y_pred):
    tp = np.sum((y == 1) & (y_pred == 1))
    tn = np.sum((y == 0) & (y_pred == 0))
    fp = np.sum((y == 0) & (y_pred == 1))
    fn = np.sum((y == 1) & (y_pred == 0))

    accuracy = (tp + tn) / (tp + tn + fp + fn)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1_score = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1_score,
        'confusion_matrix': [[tn, fp], [fn, tp]]
    }

print("=== Ex1 : Logistic Regression ===")
theta, cost_history = gradient_descent(X_train, y_train, lr=0.02, iterations=10000)
print(f"\nFinal Parameters:")
print(f"theta0 (bias): {theta[0]:.6f}")
print(f"theta1 (X1): {theta[1]:.6f}")
print(f"theta2 (X2): {theta[2]:.6f}")
Y_pred_test, Y_prob_test = predict(X_test, theta)
Y_pred_train, Y_prob_train = predict(X_train, theta)
test_metrics = evaluate_model(y_test, Y_pred_test)
train_metrics = evaluate_model(y_train, Y_pred_train)
print("\n=== Training set result ===")
print(f"Accuracy: {train_metrics['accuracy']:.4f}")
print(f"Precision: {train_metrics['precision']:.4f}")
print(f"Recall: {train_metrics['recall']:.4f}")
print(f"F1 Score: {train_metrics['f1_score']:.4f}")
print("Confusion Matrix:")
print(f"[[TN: {train_metrics['confusion_matrix'][0][0]}, FP: {train_metrics['confusion_matrix'][0][1]}], [FN: {train_metrics['confusion_matrix'][1][0]}, TP: {train_metrics['confusion_matrix'][1][1]}]]")
print("\n=== Test set result ===")
print(f"predictions: {Y_pred_test}")
print(f"Accuracy: {test_metrics['accuracy']:.4f}")
print(f"Precision: {test_metrics['precision']:.4f}")
print(f"Recall: {test_metrics['recall']:.4f}")
print(f"F1 Score: {test_metrics['f1_score']:.4f}")
print(f"Actual: {y_test}")
print("Confusion Matrix:")
print(f"[[TN: {test_metrics['confusion_matrix'][0][0]}, FP: {test_metrics['confusion_matrix'][0][1]}], [FN: {test_metrics['confusion_matrix'][1][0]}, TP: {test_metrics['confusion_matrix'][1][1]}]]")
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
train_data={
"X1":[60,62,67,70,71,72,75,78],
"X2":[22,25,24,20,15,14,14,11],
"y":[0,0,0,0,1,1,1,1]
}
test_data={
"x1":[61,71,73,79],
"X2":[23,19,15,13],
"y":[0,0,1,1]
}
df_train=pd.DataFrame(train_data)
df_test=pd.DataFrame(test_data)
x_train=df_train[["X1","X2"]].values
y_train=df_train[["y"]].values
X_test=df_test[["x1","X2"]].values
y_test=df_test[["y"]].values
model=LogisticRegression()
model.fit(x_train,y_train)
y_pred=model.predict(X_test)
accuracy=accuracy_score(y_test,y_pred)
cm=confusion_matrix(y_test,y_pred)
print("Prediction", y_pred)
print("Accuracy", accuracy)
print("Confusion Matrix",cm)
```
